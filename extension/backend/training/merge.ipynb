```python
import pandas as pd

df1 = pd.read_csv(
    "../datasets/raw/fake reviews dataset.csv"
)

df2 = pd.read_csv(
    "../datasets/raw/public_reviews_dataset_cleaned.csv",
    low_memory=False
)

print(df1.shape)
print(df2.shape)
```

```python
# Dataset 1

dataset1 = df1[
    ["text_", "label"]
].copy()

dataset1.columns = [
    "text",
    "label"
]

dataset1["label"] = dataset1[
    "label"
].map({
    "OR": 0,
    "CG": 1
})

# Dataset 2

fake_reviews = df2[
    df2["reviewer_labeled_fake"] == 1
][["review_text"]].copy()

fake_reviews["label"] = 1

genuine_reviews = df2[
    df2["reviewer_labeled_honest"] == 1
][["review_text"]].copy()

genuine_reviews["label"] = 0

dataset2 = pd.concat(
    [fake_reviews, genuine_reviews],
    ignore_index=True
)

dataset2.columns = [
    "text",
    "label"
]
```

```python
final_df = pd.concat(
    [dataset1, dataset2],
    ignore_index=True
)

final_df = final_df.dropna(
    subset=["text"]
)

final_df = final_df.drop_duplicates(
    subset=["text"]
)

final_df = final_df.sample(
    frac=1,
    random_state=42
)

print(final_df.shape)

print(
    final_df["label"]
    .value_counts()
)
```

```python
from pathlib import Path

Path("../datasets/final").mkdir(
    parents=True,
    exist_ok=True
)
```

```python
final_df.to_csv(
    "../datasets/final/deceptiscan_training.csv",
    index=False
)

print("Saved!")
```
